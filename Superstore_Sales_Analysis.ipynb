```python
df = pd.read_csv(
    r'C:\data\Sample - Superstore UTF8.csv',
    encoding='utf-8'
)
print(f"✅ Dataset loaded! Shape: {df.shape}")
print(f"   Rows: {df.shape[0]} | Columns: {df.shape[1]}")
```

```python
print("\n" + "="*55)
print("STEP 3 — FIRST LOOK AT THE DATA")
print("="*55)
 
# See first 5 rows
print("\n📋 First 5 rows:")
print(df.head())
 
# See all column names
print("\n📋 Column names:")
print(df.columns.tolist())
 
# Data types of each column
print("\n📋 Data types:")
print(df.dtypes)
 
# Basic shape info
print(f"\n📋 Total rows    : {len(df)}")
print(f"📋 Total columns : {len(df.columns)}")
```

```python
print("\n" + "="*55)
print("STEP 4 — DATA CLEANING")
print("="*55)
 
# Check for missing values
print("\n🔍 Missing values in each column:")
print(df.isnull().sum())
 
# Check for duplicate rows
duplicates = df.duplicated().sum()
print(f"\n🔍 Duplicate rows: {duplicates}")
 
# Convert Order Date and Ship Date to proper date format
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date']  = pd.to_datetime(df['Ship Date'])
 
# Extract useful date parts
df['Order Year']  = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month
df['Order Month Name'] = df['Order Date'].dt.strftime('%b')  # Jan, Feb...
df['Quarter'] = df['Order Date'].dt.quarter
 
# Calculate days to ship
df['Days to Ship'] = (df['Ship Date'] - df['Order Date']).dt.days
 
# Calculate profit margin per row
df['Profit Margin %'] = (df['Profit'] / df['Sales'] * 100).round(2)
 
print("\n✅ Data cleaning complete!")
print(f"   Date columns converted to datetime")
print(f"   New columns added: Order Year, Month, Quarter, Days to Ship, Profit Margin %")
```

```python
print("\n" + "="*55)
print("STEP 5 — BASIC STATISTICS")
print("="*55)
 
print("\n📊 Summary statistics for numeric columns:")
print(df[['Sales','Profit','Quantity','Discount']].describe().round(2))
 
total_sales   = df['Sales'].sum()
total_profit  = df['Profit'].sum()
profit_margin = (total_profit / total_sales * 100)
total_orders  = df['Order ID'].nunique()
total_customers = df['Customer ID'].nunique()
 
print(f"\n💰 Total Sales     : ${total_sales:,.2f}")
print(f"💰 Total Profit    : ${total_profit:,.2f}")
print(f"💰 Profit Margin   : {profit_margin:.2f}%")
print(f"📦 Total Orders    : {total_orders}")
print(f"👥 Total Customers : {total_customers}")
print(f"⚠️  Loss Orders     : {(df['Profit'] < 0).sum()} ({(df['Profit'] < 0).mean()*100:.1f}%)")
```

```python
# ── CHART 1: Sales by Category (Bar Chart) ───────────────────
print("\n📊 Creating Chart 1 — Sales by Category...")
 
cat_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
 
fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(cat_sales.index, cat_sales.values,
              color=['#2E75B6','#ED7D31','#70AD47'], edgecolor='white', width=0.5)
 
# Add value labels on top of each bar
for bar in bars:
    ax.text(bar.get_x() + bar.get_width()/2,
            bar.get_height() + 5000,
            f'${bar.get_height():,.0f}',
            ha='center', va='bottom', fontsize=11, fontweight='bold')
 
ax.set_title('Total Sales by Category', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Category', fontsize=12)
ax.set_ylabel('Total Sales ($)', fontsize=12)
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:,.0f}'))
ax.set_ylim(0, cat_sales.max() * 1.15)
plt.tight_layout()
plt.savefig('chart1_sales_by_category.png', bbox_inches='tight')
plt.show()
print("   ✅ Chart 1 saved as chart1_sales_by_category.png")
```

```python
# ── CHART 2: Profit by Category (Bar Chart) ──────────────────
print("\n📊 Creating Chart 2 — Profit by Category...")
 
cat_profit = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
 
fig, ax = plt.subplots(figsize=(8, 5))
colors = ['#70AD47' if v > 0 else '#FF0000' for v in cat_profit.values]
bars = ax.bar(cat_profit.index, cat_profit.values,
              color=colors, edgecolor='white', width=0.5)
 
for bar in bars:
    ypos = bar.get_height() + 1000 if bar.get_height() > 0 else bar.get_height() - 3000
    ax.text(bar.get_x() + bar.get_width()/2, ypos,
            f'${bar.get_height():,.0f}',
            ha='center', va='bottom', fontsize=11, fontweight='bold')
 
ax.set_title('Total Profit by Category', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Category', fontsize=12)
ax.set_ylabel('Total Profit ($)', fontsize=12)
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:,.0f}'))
ax.axhline(y=0, color='black', linewidth=0.8, linestyle='--')
plt.tight_layout()
plt.savefig('chart2_profit_by_category.png', bbox_inches='tight')
plt.show()
print("   ✅ Chart 2 saved as chart2_profit_by_category.png")
```

```python
# ── CHART 3: Profit by Sub-Category (Horizontal Bar) ─────────
print("\n📊 Creating Chart 3 — Profit by Sub-Category...")
 
subcat_profit = df.groupby('Sub-Category')['Profit'].sum().sort_values()
 
fig, ax = plt.subplots(figsize=(10, 7))
colors = ['#FF4444' if v < 0 else '#2E75B6' for v in subcat_profit.values]
bars = ax.barh(subcat_profit.index, subcat_profit.values,
               color=colors, edgecolor='white')
 
for bar in bars:
    width = bar.get_width()
    xpos = width + 500 if width > 0 else width - 500
    ax.text(xpos, bar.get_y() + bar.get_height()/2,
            f'${width:,.0f}',
            ha='left' if width > 0 else 'right',
            va='center', fontsize=9)
 
ax.set_title('Profit by Sub-Category\n(Red = Loss, Blue = Profit)',
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Total Profit ($)', fontsize=12)
ax.axvline(x=0, color='black', linewidth=0.8, linestyle='--')
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:,.0f}'))
plt.tight_layout()
plt.savefig('chart3_profit_by_subcategory.png', bbox_inches='tight')
plt.show()
print("   ✅ Chart 3 saved — RED bars = loss-making sub-categories!")
```

```python
# ── CHART 4: Monthly Sales Trend (Line Chart) ─────────────────
print("\n📊 Creating Chart 4 — Monthly Sales Trend...")
 
month_order = ['Jan','Feb','Mar','Apr','May','Jun',
               'Jul','Aug','Sep','Oct','Nov','Dec']
 
monthly = df.groupby('Order Month Name')['Sales'].sum()
monthly = monthly.reindex(month_order)
 
fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(monthly.index, monthly.values,
        color='#2E75B6', linewidth=2.5,
        marker='o', markersize=8, markerfacecolor='white',
        markeredgecolor='#2E75B6', markeredgewidth=2)
 
# Shade Q4 area
ax.axvspan(9.5, 11.5, alpha=0.15, color='orange', label='Q4 Peak (Nov-Dec)')
 
# Add value labels
for i, (month, val) in enumerate(zip(monthly.index, monthly.values)):
    ax.text(i, val + 3000, f'${val/1000:.0f}K',
            ha='center', va='bottom', fontsize=9, color='#2E75B6')
 
ax.set_title('Monthly Sales Trend (All Years Combined)\nQ4 spike clearly visible in Nov-Dec',
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Month', fontsize=12)
ax.set_ylabel('Total Sales ($)', fontsize=12)
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:,.0f}'))
ax.legend(fontsize=11)
plt.tight_layout()
plt.savefig('chart4_monthly_trend.png', bbox_inches='tight')
plt.show()
print("   ✅ Chart 4 saved — Q4 spike is clearly visible!")
```

```python
# ── CHART 5: Sales by Region (Pie Chart) ─────────────────────
print("\n📊 Creating Chart 5 — Sales by Region...")
 
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
 
fig, ax = plt.subplots(figsize=(8, 6))
colors = ['#2E75B6','#ED7D31','#A5A5A5','#FFC000']
wedges, texts, autotexts = ax.pie(
    region_sales.values,
    labels=region_sales.index,
    autopct='%1.1f%%',
    colors=colors,
    startangle=90,
    wedgeprops=dict(edgecolor='white', linewidth=2)
)
for text in autotexts:
    text.set_fontsize(12)
    text.set_fontweight('bold')
 
ax.set_title('Sales Distribution by Region',
             fontsize=14, fontweight='bold', pad=15)
plt.tight_layout()
plt.savefig('chart5_sales_by_region.png', bbox_inches='tight')
plt.show()
print("   ✅ Chart 5 saved as chart5_sales_by_region.png")
```

```python
# ── CHART 6: Discount vs Profit (Scatter Plot) ───────────────
print("\n📊 Creating Chart 6 — Discount vs Profit...")
 
fig, ax = plt.subplots(figsize=(9, 6))
scatter = ax.scatter(
    df['Discount'], df['Profit'],
    alpha=0.4, c=df['Profit'],
    cmap='RdYlGn', s=20
)
 
# Add trend line
z = np.polyfit(df['Discount'], df['Profit'], 1)
p = np.poly1d(z)
x_line = np.linspace(df['Discount'].min(), df['Discount'].max(), 100)
ax.plot(x_line, p(x_line), color='red', linewidth=2,
        linestyle='--', label='Trend line')
 
# Correlation value
corr = df['Discount'].corr(df['Profit'])
ax.text(0.05, 0.95, f'Correlation: {corr:.2f}',
        transform=ax.transAxes, fontsize=12,
        bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))
 
ax.axhline(y=0, color='black', linewidth=0.8, linestyle='-')
ax.axvline(x=0.2, color='orange', linewidth=1.5,
           linestyle='--', label='20% discount threshold')
 
ax.set_title('Discount vs Profit\n(Higher discounts = more losses)',
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Discount Rate', fontsize=12)
ax.set_ylabel('Profit ($)', fontsize=12)
ax.legend(fontsize=10)
plt.colorbar(scatter, ax=ax, label='Profit ($)')
plt.tight_layout()
plt.savefig('chart6_discount_vs_profit.png', bbox_inches='tight')
plt.show()
print(f"   ✅ Chart 6 saved — Correlation = {corr:.2f} (negative = bad!)")
```

```python
# ── CHART 7: Year-over-Year Sales (Bar Chart) ────────────────
print("\n📊 Creating Chart 7 — Year-over-Year Sales...")
 
yearly = df.groupby('Order Year').agg(
    Sales=('Sales','sum'),
    Profit=('Profit','sum')
).reset_index()
 
x = np.arange(len(yearly))
width = 0.35
 
fig, ax = plt.subplots(figsize=(9, 6))
bars1 = ax.bar(x - width/2, yearly['Sales'],   width, label='Sales',
               color='#2E75B6', edgecolor='white')
bars2 = ax.bar(x + width/2, yearly['Profit'],  width, label='Profit',
               color='#70AD47', edgecolor='white')
 
for bar in bars1:
    ax.text(bar.get_x() + bar.get_width()/2,
            bar.get_height() + 5000,
            f'${bar.get_height()/1000:.0f}K',
            ha='center', va='bottom', fontsize=10, fontweight='bold')
for bar in bars2:
    ax.text(bar.get_x() + bar.get_width()/2,
            bar.get_height() + 2000,
            f'${bar.get_height()/1000:.0f}K',
            ha='center', va='bottom', fontsize=10, fontweight='bold',
            color='#70AD47')
 
ax.set_title('Year-over-Year Sales & Profit',
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Year', fontsize=12)
ax.set_ylabel('Amount ($)', fontsize=12)
ax.set_xticks(x)
ax.set_xticklabels(yearly['Order Year'])
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:,.0f}'))
ax.legend(fontsize=11)
plt.tight_layout()
plt.savefig('chart7_yearly_sales.png', bbox_inches='tight')
plt.show()
print("   ✅ Chart 7 saved as chart7_yearly_sales.png")
```

```python
# ── CHART 8: Correlation Heatmap ─────────────────────────────
print("\n📊 Creating Chart 8 — Correlation Heatmap...")
 
numeric_cols = df[['Sales','Profit','Quantity','Discount','Days to Ship']]
corr_matrix  = numeric_cols.corr().round(2)
 
fig, ax = plt.subplots(figsize=(8, 6))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=1)
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    linewidths=0.5,
    linecolor='white',
    ax=ax,
    annot_kws={'size': 12}
)
ax.set_title('Correlation Heatmap — Numeric Columns\n'
             '(+1 = strong positive, -1 = strong negative)',
             fontsize=13, fontweight='bold', pad=15)
plt.tight_layout()
plt.savefig('chart8_correlation_heatmap.png', bbox_inches='tight')
plt.show()
print("   ✅ Chart 8 saved — Key finding: Discount & Profit are negatively correlated!")
```
